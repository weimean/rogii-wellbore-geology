# 01 — EDA: ROGII Wellbore Geology Prediction

Goal: understand one well before modelling many.

1. Load a single well (horizontal + typewell).
2. Plot `GR` vs `MD` for the horizontal well; mark the **evaluation zone** (`TVT_input` is NaN).
3. Plot the typewell `GR` vs `TVT` with `Geology` labels.
4. Eyeball how the horizontal `GR` shape maps onto the typewell signature — that mapping *is* the target `TVT`.

See `04 Projects/Kaggle — ROGII Wellbore Geology Prediction.md` in the vault for the running decision log.

```python
from pathlib import Path
import matplotlib.pyplot as plt

import sys; sys.path.append('..')
from src.data import load_split, TARGET, TARGET_INPUT, LOG

wells = load_split(Path('../data/raw/train'), limit=1)
w = wells[0]
print(w.name, 'eval-zone rows:', int(w.eval_mask.sum()), '/', len(w.horizontal))
w.horizontal.head()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(11, 5))
ax[0].plot(w.horizontal[LOG], w.horizontal['MD']); ax[0].invert_yaxis()
ax[0].set(title=f'Horizontal GR — {w.name}', xlabel='GR', ylabel='MD')
ax[1].plot(w.typewell[LOG], w.typewell[TARGET]); ax[1].invert_yaxis()
ax[1].set(title='Typewell GR', xlabel='GR', ylabel='TVT')
plt.tight_layout()
```
